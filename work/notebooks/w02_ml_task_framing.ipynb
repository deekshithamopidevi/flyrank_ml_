## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

For example, a page can have high impressions but falling clicks, low traffic but poor position, or strong recent performance with other signs of decline. These signals can interact in ways that are difficult to capture with one simple if-statement. ML can learn patterns from several measured features and produce a ranking that helps prioritize pages for review. The result should be treated as decision-support, not as proof that a page will definitely decline.
